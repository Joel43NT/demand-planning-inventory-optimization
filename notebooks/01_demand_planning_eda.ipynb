```python
import numpy as np
import pandas as pd

np.random.seed(42)

# Configuración del dataset
n_skus = 50
weeks = 104  # 2 años de historia semanal (2024-2025)

date_range = pd.date_range(start="2024-01-01", periods=weeks, freq="W-MON")

skus = [f"SKU_{str(i).zfill(3)}" for i in range(1, n_skus + 1)]
categories = np.random.choice(
    ["Electronica", "Perecederos", "Industrial", "Moda"], size=n_skus
)

data = []

for i, sku in enumerate(skus):
  cat = categories[i]
  base_demand = np.random.randint(100, 2000)
  seasonality_factor = np.random.uniform(0.1, 0.5)
  unit_cost = np.random.uniform(10, 500)
  lead_time = np.random.choice([1, 2, 4, 8])  # semanas de lead time
  moq = np.random.choice([50, 100, 500])  # Minimum Order Quantity

  for w, date in enumerate(date_range):
    # Simular estacionalidad (pico fuerte en semanas de invierno / fin de año)
    seasonal_spike = (
        1
        + seasonality_factor
        * np.sin(2 * np.pi * w / 52)
        * (1 if (w % 52 > 40 or w % 52 < 5) else 0.3)
    )
    noise = np.random.normal(0, 0.15)
    actual_demand = max(
        10, int(base_demand * seasonal_spike * (1 + noise))
    )

    data.append({
        "sku_id": sku,
        "category": cat,
        "date": date,
        "weekly_demand": actual_demand,
        "unit_cost": round(unit_cost, 2),
        "lead_time_weeks": lead_time,
        "moq": moq,
        "holding_cost_rate": 0.18,  # 18% anual del costo de inventario
    })

df_demand = pd.DataFrame(data)

# Guardar en la carpeta data (subiendo un nivel desde notebooks)
df_demand.to_csv("../data/demand_inventory_data.csv", index=False)
print(
    "Dataset de Planeación de Demanda generado con éxito:"
    f" {df_demand.shape[0]} registros guardados en /data."
)
```
